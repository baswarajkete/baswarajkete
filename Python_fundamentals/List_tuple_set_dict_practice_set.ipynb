```python
# Find input
def find_max(l):
    maximum = None

    for x in l:
        if isinstance(x, list):
            value = find_max(x)
        else:
            value = x

        if maximum is None or value > maximum:
            maximum = value

    return maximum
find_max(l)
```

```python
#Using dictionary comprehension, create a dictionary where:

#keys remain unchanged
#values greater than 15 are doubled
#values 15 or below remain unchanged

d = {
    "apple": 10,
    "banana": 25,
    "orange": 15,
    "grapes": 30
}

d2 = {key : value*2 if value>15 else value for key,value in d.items() }
print(d2)
```

```python
#Find the student who has the highest combined marks in Math + Science.

students = {
    "s1": {"name": "Raj", "marks": {"math": 80, "science": 75}},
    "s2": {"name": "Amit", "marks": {"math": 90, "science": 85}},
    "s3": {"name": "John", "marks": {"math": 70, "science": 95}}
}
max_marks=0
ranker=None
for key, value in students.items():

    name = value["name"]
    marks = value["marks"]

    math = marks["math"]
    science = marks["science"]

    total = math + science
    if total> max_marks:
      max_marks = total
      ranker = name


print(ranker, max_marks)
```

```python
#Create a new dictionary containing only the entries where:

#the value is greater than 10
#the value is even

d = {
    "a": 5,
    "b": 12,
    "c": 7,
    "d": 20,
    "e": 15
}

d2 = {x : y for x,y in d.items() if y>10 and y%2==0}
print(d2)
```

```python
#Find the person with the highest average score across all teams.

data = {
    "team1": {
        "Raj": [80, 90, 70],
        "Amit": [60, 75, 85]
    },
    "team2": {
        "John": [95, 85, 90],
        "Sam": [70, 80, 75]
    }
}
total=0
max1=0
ranker = None
for member in data.values():
  #print(sum(member.values()))
  for name,score in member.items():
    total=sum(score)/len(score)
    if total > max1:
      max1 = total
      ranker = name

print(ranker,max1)

```

```python
#Create a new dictionary where:

#duplicate values are grouped together
#each value becomes a key
#the corresponding value is a list of keys having that value

d = {
    "A": 10,
    "B": 20,
    "C": 10,
    "D": 30,
    "E": 20,
    "F": 40
}

d2 ={}
for key, value in d.items():
    if value not in d2:
        d2[value] = [key]
    else:
        d2[value].append(key)

print(d2)
```
