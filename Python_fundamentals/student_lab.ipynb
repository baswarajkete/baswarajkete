## 📌 Exercise 1: Build a `Car` Class

Create a class named `Car` that:
1. Takes `brand`, `model`, and `price` as parameters in `__init__`.
2. Has an instance variable `speed` initialized to `0`.
3. Has a method `accelerate(amount)` that increases `speed` by `amount` and prints the current speed.
4. Has a method `display_info()` that prints brand, model, price, and current speed.

```python
class Car:
    def __init__(self, brand: str, model: str, price: float):
        # TODO: Initialize instance variables
        self.brand = brand
        self.model = model
        self.price = price
        self.speed = 0

    def accelerate(self, amount: int):
        # TODO: Increase speed and print current speed
        self.speed += amount
        print(self.speed)

    def display_info(self):
        # TODO: Print car details
        print(f"Brand: {self.brand}\nModel: {self.model}\nPrice: {self.price}\nCurrent Speed: {self.speed}")

# Test your Car class:
my_car = Car("Tata", "Nexon", 1200000)
my_car.display_info()
my_car.accelerate(40)
my_car.display_info()

```

## 📌 Exercise 2: Build a `BankAccount` System

Create a class `BankAccount` that:
1. Takes `account_holder` and initial `balance` in `__init__`.
2. Has a `deposit(amount)` method that adds `amount` to `balance`.
3. Has a `withdraw(amount)` method that subtracts `amount` from `balance` (if sufficient funds exist).
4. Has a `get_balance()` method that returns the current balance.

```python
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0):
        # TODO: Initialize attributes
        self.balance = balance

    def deposit(self, amount: float):
        # TODO: Deposit logic
        self.balance +=amount

    def withdraw(self, amount: float):
        # TODO: Withdraw logic with check
        if self.balance>= amount:
           self.balance-= amount
        else:
            print("Insufficient funds")

    def get_balance(self) -> float:
        # TODO: Return balance
        return self.balance

# Test your BankAccount class:
acc = BankAccount("Priya", 5000)
acc.deposit(2500)
acc.withdraw(1000)
print(acc.get_balance())
```

## 📌 Exercise 3: Instance vs. Class Variables

Create a class `StudentTracker` that:
1. Has a class variable `total_students = 0`.
2. In `__init__`, takes `name` and `grade`, and increments `StudentTracker.total_students` by `1` every time a new student object is created.
3. Has a method `get_student_count()` that prints the total number of students created so far.

```python
class StudentTracker:
    total_students = 0  # Class variable

    def __init__(self, name: str, grade: str):
        # TODO: Initialize instance attributes and increment total_students
        self.name = name
        self.grade = grade
        StudentTracker.total_students +=1

    @classmethod
    def get_student_count(cls):
        print(f"Total Students Registered: {cls.total_students}")

# Test your StudentTracker class:
s1 = StudentTracker("Aarav", "A")
s2 = StudentTracker("Diya", "B")
StudentTracker.get_student_count()  # Should output: 2
```
