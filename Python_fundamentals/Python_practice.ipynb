```python
#Create a BankAccount class with an account holder and balance.
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance

    def __add__(self,other):
        return self.balance + other.balance
    def __str__(self):
        return self.owner, self.balance

a1 = BankAccount("Rahul", 5000)
a2 = BankAccount("Priya", 7000)

print(a1 + a2)
```

```python
#Create a parent class Employee with a method calculate_salary() that returns the base salary.

class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def calculate_salary(self):
        return self.salary

class Manager(Employee):
    def __init__(self,name,salary,bonus):
        super().__init__(name,salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary+ self.bonus


manager = Manager("Rahul", 50000, 10000)
print(manager.calculate_salary())
```

```python
data = {
    'a': {
        'b': {
            'c': 10
        }
    },
    'd': 20
}

def flatten(d, prefix=''):
    result = {}

    for key, value in d.items():
        new_key = f"{prefix}.{key}" if prefix else key

        if isinstance(value, dict):
            nested = flatten(value, new_key)
            result.update(nested)
        else:
            result[new_key] = value

    return result


print(flatten(data))
```

```python
#12.A Write a function using *args to find the largest numeric argument without using max().
def largest_num(*args):
    largest = float('-inf')
    if not args:
        raise ValueError("At least one number is required")

    for i in args:
        if i > largest:
            largest=i
    return largest

print(largest_num(10,20,30,10))
print(largest_num())

```

```python
#16.B Write a decorator that measures execution time for a function summing integers from 1 to 10000.
import time

def timer(func):
    def wrapper(*args):
        start = time.perf_counter()
        result = func(*args)
        print("Execution time by",func.__name__,time.perf_counter()-start,'sec')
        return result
    return wrapper
@timer
def sum_of_integers(n):
    total=0
    for i in range(1,n):
        total +=i
    print(total)

sum_of_integers(10000)
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Rectangle(Shape):

    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth


c = Circle(3)
r = Rectangle(4, 5)

print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

```python
#25.A Convert a 12-hour time string such as "07:45 PM" into 24-hour format without using datetime.
time = "07:45 PM"

hour, remaining = time.split(':')
minute, period = remaining.split()

hour = int(hour)

if period == 'AM':
    if hour == 12:
        hour = 0
else:
    if hour != 12:
        hour += 12

result = f"{hour:02d}:{minute}"

print(result)
```
