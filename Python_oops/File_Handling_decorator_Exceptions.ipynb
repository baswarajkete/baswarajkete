Decorator

```python
def my_decorator(func):
    def wrapper():
        print("---------------")
        func()
        print("---------------")
    return wrapper

@my_decorator
def hello():
    print("Heloo..")


hello()
```

```python
#to calculate the time taking of fucntions
import time

def timer(func):
    def wrapper(*args):
        start = time.time()
        func(*args)
        print('Time taken by',func.__name__,time.time()-start,'secs')
    return wrapper

@timer
def hello():
    time.sleep(2)
    print("hello world")

@timer
def square(num):
    print(num**2)

@timer
def power(a,b):
    print(a**b)

hello()
square(2)
power(2,3)

```

```python
def sanity_check(datatype):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if all(type(arg)==datatype for arg in args):
                func(*args)
            else:
                raise TypeError('Invalid data type')
        return inner_wrapper
    return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

@sanity_check(int)
def power(a,b):
    print(a**b)

square(2)
greet("Ram")
power(2,3)
```

```python
class MyException(Exception):
    pass

class BankAccount():
    def __init__(self,name,balance):
        self.name = name
        if balance<0:
            raise MyException("Balance should not be negative")
        self.balance= balance

    def withdraw(self,amount):
        if amount<0:
            raise MyException("Amount should not be negative")
        if self.balance < amount:
            raise MyException("Insufficient funds")
        self.balance -= amount


try:
    p = BankAccount('Ram', 1000)
    p.withdraw(-100)
    p.withdraw(500)

except MyException as e:
    print(e)

print(p.balance)
```
