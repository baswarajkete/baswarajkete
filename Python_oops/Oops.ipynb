```python
class Fractions:
    def __init__(self, x,y):
        self.x = x
        self.y = y
    def __str__(self):
        return '{}/{}'.format(self.x,self.y)

    def __add__(self,other):
        new_num = self.x * other.y + self.y*other.x
        new_din = self.y*other.y
        return '{}/{}'.format(new_num,new_din)

    def __sub__(self,other):
        new_num = self.x * other.y - self.y*other.x
        new_din = self.y*other.y
        return '{}/{}'.format(new_num,new_din)

    def __truediv__(self,other):
        new_num = self.x * other.y
        new_din = self.y*other.x
        return '{}/{}'.format(new_num,new_din)
```

Create coordinates class
- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line


```python
class Point:
    def __init__(self,x, y):
        self.x_cod = x
        self.y_cod = y

    def euclidean_distance(self,other):
        return ((self.x_cod-other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

    def center_distance(self):
        return self.euclidean_distance(Point(0,0))

    def __eq__(self,other):
        return self.x_cod == other.x_cod and self.y_cod == other.y_cod

class Line:
    def __init__(self,A,B,C):
        self.a = A
        self.b = B
        self.c = C

    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.a,self.b,self.c)

    def pointonline(self,point):
        if self.a*point.x_cod + self.b*point.y_cod + self.c ==0:
            return "Point is on the line"
        else:
            return "Point is not on the line"

    def __contains__(self,point):
        return self.a*point.x_cod + self.b*point.y_cod + self.c ==0



p1 = Point(2,2)
p2 = Point(1,1)
print(p1.euclidean_distance(p2))
print(p1.center_distance())

l1 = Line(1,2,3)
print(l1)
print(l1.pointonline(p1))
print(l1.pointonline(p2))
p3 = Point(-1,-1)
print(l1.pointonline(p3))

print(p3 in l1)
print(p1==p2)
```

Project: Bank Management System

```python
class BankAccount:
    def __init__(self,account_holder,balance=0):
        self.account_holder = account_holder
        if balance >= 0:
           self.__balance = balance
        else:
            print("Invalid balace input")
        self.__transactions=[]

    def get_balance(self):
        return self.__balance

    def __str__(self):
        return f'Account holder: {self.account_holder}'

    def deposit(self,amount):
        if amount>0:
            self.__balance += amount
            print("Deposited:", amount)
            self.__transactions.append(f"Deposited: {amount}")
        else:
            print("Invalid input amount")

    def withdraw(self,amount):
        if amount >0:
            if self.__balance >= amount:
                self.__balance -= amount
                print("Withdrawn:",amount)
                self.__transactions.append(f"Withdrawn: {amount}")
            else:
                print("Insufficient funds")
        else:
            print("Invalid input amount")

    def _record_withdrawal(self,amount):
        self.__balance -= amount
        self.__transactions.append(f"Withdrawn: {amount}")
        print("Withdrawn:", amount)

    def check_balance(self):
        print (f'Current balance: {self.__balance}')

    def get_transactions(self):
        return self.__transactions.copy()

    def show_transactions(self):
        print("Transactions:")
        for i in self.get_transactions():
            print(i)


class SavingsAccount(BankAccount):
    def __init__(self,account_holder,balance,interest_rate=5):
        super().__init__(account_holder,balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        amount = self.get_balance() * self.interest_rate/100
        super().deposit(amount)

    def show_account_details(self):
        print(f"Account holder: {self.account_holder} Interest rate: {self.interest_rate}")



class CurrentAccount(BankAccount):
    def __init__(self,account_holder, balance, overdraft_limit=5000):
        super().__init__(account_holder, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if amount>0:
            if self.get_balance() + self.overdraft_limit >= amount:
                self._record_withdrawal(amount)
            else:
                print("Crossed overdraft_limit amount")
        else:
            print("Enter positive amount")


p = BankAccount("Ram")
print(p)
p.deposit(1000)
p.withdraw(200)
p.check_balance()
p.show_transactions()

s = SavingsAccount("Ram", 1000, 5)
s.add_interest()
s.check_balance()
s.show_account_details()

s.show_transactions()

c = CurrentAccount('sham',1000)
c.withdraw(4000)
c.check_balance()
c.withdraw(1000)
c.check_balance()
c.withdraw(1000)
c.check_balance()
c.show_transactions()

accounts = [s, c]

for account in accounts:
    print(account.account_holder)
    account.withdraw(500)
    account.check_balance()


```
